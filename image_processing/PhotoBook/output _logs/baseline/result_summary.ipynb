```python
import json
import re
```

```python
def delete_zero(s: str) -> str:
    return re.sub(r'^0+', '', s)
```

```python
FILE_NAME = "baseline_qwen2_5_vl_2026-10-23 15:53:44.345763.json"
```

```python
with open(FILE_NAME, "r") as file:
    data = json.load(file)

    for i, case in enumerate(data):
        print(f"\n\nANSWER {case}")
        print(data[case][0][0])

        correct = data[case][1][3]

        
        print("\nCORRECT")
        counter = 0
        nmbr2titel = {}
        titel = ''
        og_nmbr = ''
        for key in data[case][0][1][1]['content']:
            if counter < 12:
                if key['type'] == 'text':
                    titel = key['text']
                    # print(titel)
                elif key['type'] == 'image':
                    og_nmbr = delete_zero(key['image'][-10:-4])
                else:
                    print('error')
                    
                if counter % 2 != 0:
                    nmbr2titel[og_nmbr] = titel
                counter += 1
                
        # print(nmbr2titel)
        print(nmbr2titel[str(correct)])

        # print("\nCORRECT")
        print(correct)
        print("----------------------------------------------------------")


```
