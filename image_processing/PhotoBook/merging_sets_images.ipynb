```python
from PIL import Image
from matplotlib import pyplot as plt
```

```python
def merge2images(img_path1, img_path2, out_path, line_width=20, line_color=(255, 255, 255)):

    im1 = Image.open(img_path1)
    im2 = Image.open(img_path2)

    min_h = min(im1.height, im2.height)
    min_w = min(im1.width, im2.width)

    # im1 = im1.resize((int(im1.width * min_h / im1.height), min_h))
    # im2 = im2.resize((int(im2.width * min_h / im2.height), min_h))

    im1 = im1.resize((min_w, int(im1.height * min_w / im1.width)))
    im2 = im2.resize((min_w, int(im2.height * min_w / im2.width)))

    # new_w = im1.width + line_width + im2.width
    # new_h = min_h
    new_w = min_w
    new_h = im1.height + line_width + im2.height
    canvas = Image.new("RGB", (new_w, new_h), (255, 255, 255))

    canvas.paste(im1, (0, 0))

    # line_img = Image.new("RGB", (line_width, new_h), (0,0,0))
    # canvas.paste(line_img, (im1.width, 0))

    line_img = Image.new("RGB", (new_w, line_width), line_color)
    canvas.paste(line_img, (0, im1.height))


    # canvas.paste(im2, (0, im1.width + line_width, 0))
    canvas.paste(im2, (0, im1.height + line_width))

    canvas.save(out_path)

merge2images("sets_images_stripe/2106_2_A.jpg", "sets_images_stripe/2106_2_B.jpg", "merged_images/2106_2_merged.jpg")
```
